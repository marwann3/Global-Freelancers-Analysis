```python
import pandas as pd
```

```python
df = pd.read_csv('/content/archive.zip')
```

```python
df.head()
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
df.rename(columns={'is_active': 'active_status'}, inplace=True)
```

```python
print(df.columns)
```

```python
df.describe()
```

```python
df.info()
```

```python
df.drop('hourly_rate (USD)', axis=1, inplace=True)
```

```python
df.duplicated().sum()
```

```python
null_percent = (df.isnull().sum() / len(df)) * 100
null_percent.sort_values(ascending=False)
```

```python
df['gender'] = df['gender'].replace({
    'F': 'Female',
    'f': 'Female',
    'FEMALE': 'Female',
    'female': 'Female',
    'M': 'Male',
    'm': 'Male',
    'MALE': 'Male',
    'male': 'Male'
})
```

```python
df['gender'].value_counts(dropna=False)
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['age'].isnull().sum()
```

```python
df['years_of_experience'] = df['years_of_experience'].fillna(
    df['years_of_experience'].median()
)
```

```python
df['years_of_experience'].isnull().sum()
```

```python
df['rating'] = df['rating'].fillna(df['rating'].median())
```

```python
df['rating'].isnull().sum()
```

```python
df['active_status'] = df['active_status'].replace({
    0: False,
    '0': False,
    'N': False,
    'n': False,
    'No': False,
    'NO': False,
    'no': False,
    'FALSE': False,
    'False': False,
    'false': False,

    1: True,
    '1': True,
    'Y': True,
    'y': True,
    'Yes': True,
    'YES': True,
    'yes': True,
    'TRUE': True,
    'True': True,
    'true': True
})
```

```python
df['active_status'].value_counts(dropna=False)
```

```python
df['active_status'] = df['active_status'].fillna('Unknown')
```

```python
df['active_status'].isnull().sum()
```

```python
# إزالة علامة %
df['client_satisfaction'] = df['client_satisfaction'].str.replace('%', '', regex=False)

# تحويل العمود لرقمي
df['client_satisfaction'] = pd.to_numeric(df['client_satisfaction'], errors='coerce')

# تعويض الـ Nulls بالـ Median
df['client_satisfaction'] = df['client_satisfaction'].fillna(
    df['client_satisfaction'].median()
)

# تحويله لـ int
df['client_satisfaction'] = df['client_satisfaction'].astype(int)
```

```python
df['client_satisfaction'].isnull().sum()
```

```python
df[['age', 'years_of_experience']] = (
    df[['age', 'years_of_experience']]
    .astype(int)
)
```

```python
df.info()
```

```python
df.to_csv('cleaned_data.csv', index=False)
```
